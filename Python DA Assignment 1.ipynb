### <b> <span style="color:darkblue">Pandas Series: </span> </b>

<b> <span style="color:purple">1. Creating Pandas Series:  </span></b>

<b>● Create a Pandas Series named marks with the following data:</b>

<b>➢ Marks: 95, 92, 89, 85, 80 </b>

<b>➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'</b>

```python
import pandas as pd
```

```python
Marks=pd.Series([95,92,89,85,80],index=['Rank1','Rank2','Rank3','Rank4','Rank5'])
print(Marks)
```

<span style="color:green"><b>Pandas is imported for creating Series.</b></span>

<span style="color:green"><b>The pd.Series() function creates a Series named marks with five marks and custom indices.</b></span>

<b> <span style="color:purple">2. Indexing and Slicing:  </span></b>

<b>● Use the integer index position to access the mark of the 1st rank student. </b>

```python
print("The mark of the 1st rank student : ",Marks.iloc[0])
```

<span style="color:green"><b>iloc is used to acess the mark of 1st rank student.iloc accessor retrieves values using integer positions.</b></span>

<b>● Use the loc accessor to retrieve the marks of the top 3 ranks by 
specifying their index labels. </b>

```python
print(Marks.loc['Rank1':'Rank3'])
```

<b><span style="color:green">The loc accessor selects values using their index labels.</span></b>

<b>● Use the iloc accessor to retrieve the mark of the 3rd rank student. </b>

```python
print("The mark of the 3rd rank student : ",Marks.iloc[2])
```

<b>● Apply a boolean mask to filter the Series and retrieve the ranks where the 
marks are greater than 90. </b>

```python
print(Marks[Marks>90])
```

<span style="color:green"><b>A Boolean mask checks which marks are greater than 90.Rank1 and Rank2 has mark greater than 90.</b></span>

<b> <span style="color:purple">3. Manipulating Series:  </span></b>

<b>● Modify the mark of the student with the 1st rank to 100.</b>

```python
Marks['Rank1']=100
print(Marks)
```

<span style="color:green"><b>The statement marks['Rank1'] = 100 updates the mark associated with Rank1 from 95 to 100.</b></span>

<b>● Remove the entry corresponding to the last rank from the Series. </b>

```python
Marks=Marks.drop(['Rank5'])
print(Marks)
```

<span style="color:green"><b>The drop() method removes the entry with the index Rank5.</b></span>

<b>● Compute the CGPA by dividing each mark in the Series by 10.</b>

```python
CGPA=Marks/10
print(CGPA)
```

### <b> <span style="color:darkblue">Pandas DataFrame: </span> </b>

<b> <span style="color:purple">1. Creating Pandas DataFrame:  </span></b>

<b>● Create a Pandas DataFrame named transactions with the following data: </b>

<b>➢ TransactionID: 101, 102, 103, 104, 105, 106, 107, 108, 109, 110 </b>

<b>➢ ProductCategory: 'Electronics', 'Clothing', 'Electronics', 'Furniture', 
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'</b>

<b>➢ Region: 'North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 
'South', 'North' </b>

<b>➢ Amount: 200, 150, 300, 450, 200, 250, 300, 180, 350, 400</b>

```python
print(transactions.iloc[:,-3:])
```

<span style="color:green"><b>The iloc accessor selects data by integer position. The colon : selects all rows and -3: selects the last three columns.</b></span>

<b>● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.</b>

```python
print(transactions[(transactions['Region']=='North')&(transactions['Amount']>200)])
```

<span style="color:green"><b>Boolean filtering selects rows that satisfy both conditions.<b></span>

<b>● Find the value counts for the 'ProductCategory' column. </b>

```python
print(transactions.value_counts("ProductCategory"))
```

<span style="color:green"><b>The value_counts() method counts how many times each product category appears.</b></span>

<span style="color:green"><b>The unique() method returns the distinct values in the column Region.</b></span>

<b>● Group by 'Region' and find the mean amount for each region.</b>

```python
transactions.groupby("Region")["Amount"].mean()
```

<span style="color:green"><b>The groupby() method groups transactions by region, and mean() calculates the average transaction amount for each region.</b></span>

<b> <span style="color:purple">3. Manipulating the DataFrame: </span></b>

<b>● Modify the 'Amount' for the transaction with TransactionID 102 to 165.</b>

```python
transactions.loc[transactions["TransactionID"]==102,"Amount"]=165
print(transactions)
```

<span style="color:green"><b>The loc accessor selects the row where TransactionID is 102 and updates its Amount from 150 to 165.</b></span>

<b>● Add a new column 'Discount' by calculating 10% of the 'Amount'.</b>

```python
transactions["Discount"]=transactions["Amount"]*10/100
print(transactions)
```

<b>● Remove the row with TransactionID 109. </b>

```python
transactions=transactions.drop(transactions[transactions["TransactionID"]==109].index)
print(transactions)
```

<span style="color:green"><b>The .index finds the index of the identified row, and the drop() method removes that row from the DataFrame.</b></span>

<b>● Delete the 'Discount' column from the DataFrame.</b>

```python
transactions.drop(columns=["Discount"],inplace=True)
print(transactions)
```
